Overfitting tab hoti hai jab model training data ko bohat zyada yaad kar leta hai, lekin naye unseen data par acha perform nahi karta.

Sab se simple example

Suppose aap ke paas 100 cat/dog images hain.

Model training ke dauran ye 100 images itni achi tarah memorize kar leta hai ke training accuracy:

$$ 99\% $$

ho jati hai.

Lekin jab aap usay new images dete ho:

$$ Validation/Test Accuracy = 75\% $$

To iska matlab:

Model ne general rule nahi seekha, balki training examples ki details memorize kar li.

Yahi overfitting hai.

Image ke left graph ka meaning

Left side mein blue straight-ish line general trend hai, aur purple/red curve har training point ko follow karne ki koshish kar rahi hai.

Simple idea:

Correct learning:
Data ka overall pattern seekho

Overfitting:
Har choti noise / unnecessary detail bhi memorize karo

Model actual pattern ke saath-saath noise bhi learn kar leta hai.

Causes of Overfitting

Image mein 2 causes likhe hain.

1. **Too much parameter freedom**

Agar model bohat bada hai:

Too many layers

Too many neurons

Too many weights

aur data kam hai, to model ke paas itni flexibility hoti hai ke wo training data ko memorize kar sakta hai.

**Example:**

Training examples = 500

Parameters = 5 million

Model ke paas bohat zyada freedom hai.

2. **Not enough data**

Agar training data kam ho:

50 images

to model ko real-world variation kam dekhne ko milti hai.

Wo soch sakta hai:

"Jo maine training mein dekha hai, wahi duniya hai."

Lekin new data different ho sakta hai.

Effect kya hota hai?

Image mein:

High training-set specificity

Yani model training set ke liye bohat specially tuned ho jata hai.

Training data → excellent

New data      → poor


**Poor generalization**

Generalization ka matlab:

Jo model ne training mein seekha hai, kya wo naye unseen examples par bhi kaam karta hai?

Overfitting mein:

$$ \boxed{\text{Training performance high}} $$

lekin:

$$ \boxed{\text{Validation/Test performance poor}} $$
Ab right side ka Loss graph

Ye overfitting detect karne ka bohat important graph hai.

X-axis:

$$ Epochs $$

Y-axis:

$$ Loss $$

Initially:

Training Loss   ↓

Validation Loss ↓

Dono improve kar rahe hote hain.

Phir ek point ke baad:

Training Loss   ↓ ↓ ↓

Validation Loss ↑

Yahan overfitting start ho gayi.

Matlab:

Model training data par ab bhi better ho raha hai, lekin validation data par worse ho raha hai.

Ye bohat classic sign hai.
