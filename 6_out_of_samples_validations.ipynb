Hyperparameters woh settings hoti hain jo hum model training se pehle choose karte hain, aur training unko automatically update nahi karti.

Example:

Step size / Learning rate → model har update mein kitna bada step le.
Iterations / Epochs → model training data ko kitni baar process kare.

Aur bhi hyperparameters ho sakte hain, jaise batch size, number of layers, number of neurons, dropout, etc.

Sab se important difference:

**Model Parameters**
→ training ke dauran model khud learn karta hai
Example: weights, biases

**Hyperparameters**
→ hum pehle set/tune karte hain
Example: learning rate, epochs, batch size

Simple example:

Agar aap student ko train kar rahe ho:

Hyperparameter: “Roz 2 hours study karo, 10 practice tests do.”
Parameter: Student ne practice se actual knowledge kitni learn ki.

So:

Parameters = model kya seekhta hai
Hyperparameters = model kaise seekhega
