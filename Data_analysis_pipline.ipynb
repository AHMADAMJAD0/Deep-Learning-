# **ML data pipline**

1. **Data collection**

Machine learning model data se learn karta hai. Is liye normally pipeline ka pehla step data collect karna hota hai.

Agar aap new deep-learning problem solve karna chahte hain, to sabse pehle yeh decide karna hoga:

Model ko train karne ke liye data kahan se aur kis form mein milega?

Example:

Agar hum cats aur dogs ko classify karna chahte hain, to humein cats aur dogs ki labeled images collect karni hongi.

2. **Enough data**

Model ke paas patterns seekhne ke liye sufficient examples hone chahiye.

Example:

Agar hum model ko sirf:

5 cat images

5 dog images

dikhayen, to model reliable patterns nahi seekhega.

Deep learning ki recent success ka ek major reason yeh hai ke models ko extremely large datasets par train kiya gaya hai.

Simple idea:

Kam data → limited learning

Zyada useful data → better pattern learning

Lekin sirf data quantity important nahi; data quality aur variety bhi important hai.
